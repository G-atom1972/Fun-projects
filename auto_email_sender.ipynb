```python
import smtplib
import csv
import time
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart

# --- CONFIG ---
SMTP_SERVER = "smtp.gmail.com"   # change if not using Gmail
SMTP_PORT = 587
SENDER_EMAIL = "helloworld@gmail.com"
SENDER_PASSWORD = "app-password"   # use an App Password, not your real password

SUBJECT_TEMPLATE = "MUN Allotment Confirmation - {committee}"
BODY_TEMPLATE = """Dear {name},

Congratulations! You have been allotted the portfolio of {allotment} in the {committee} committee.

Please confirm your participation at your earliest convenience.

Best regards,
Secretariat
"""

def send_emails(csv_path):
    server = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
    server.starttls()
    server.login(SENDER_EMAIL, SENDER_PASSWORD)

    with open(csv_path, newline='', encoding='utf-8-sig') as f:  # <-- utf-8-sig strips BOM automatically
        reader = csv.DictReader(f)
        reader.fieldnames = [h.strip().lower() for h in reader.fieldnames]

        for row in reader:
            row = {k.strip().lower(): (v.strip() if v else v) for k, v in row.items()}

            msg = MIMEMultipart()
            msg['From'] = SENDER_EMAIL
            msg['To'] = row['email']
            msg['Subject'] = SUBJECT_TEMPLATE.format(**row)

            body = BODY_TEMPLATE.format(**row)
            msg.attach(MIMEText(body, 'plain'))

            server.sendmail(SENDER_EMAIL, row['email'], msg.as_string())
            print(f"Sent to {row['name']} ({row['email']})")

            time.sleep(1)

    server.quit()
    print("All emails sent!")

if __name__ == "__main__":
    send_emails("Book1.csv")
```
