```python
#thought process
# normal vigenere cipher shifts, can be done using ascii but it also has to loop around

str1=input("enter a string ")
shift=int(input("enter shift value "))
for i in str1:
  v_str=ord(i)
  v_str=ord(i)+shift
  print(chr(v_str),end="")
```

```python
#basic now works lets loop it around
#currently a caesar cipher
"""ord('y') - ord('a')   # 24
24 + 3                # 27
27 % 26               # 1
1 + ord('a')          # ASCII for 'b'"""


result=""
str1=input("enter a string ")
shift=int(input("enter shift value "))
for i in str1:
  if i.isalpha():
    new_char=chr((ord(i)-ord("a")+shift)%26+ord("a"))
    result+=new_char
  elif i.isspace():
    result+=" "
  else:
    result+=new_char
print(result)

```

```python
#aes implementation


from Crypto.Cipher import AES
from secrets import token_bytes

key=token_bytes(16)

def encrypt(msg):
  cipher= AES.new(key,AES.MODE_EAX)
  nonce=cipher.nonce
  ciphertext, tag= cipher.encrypt_and_digest(msg.encode("ascii"))
  return nonce,ciphertext,tag

def decrpyt(nonce,ciphertext,tag):
  cipher=AES.new(key,AES.MODE_EAX,nonce=nonce)
  plaintext=cipher.decrypt(ciphertext)
  try:
    cipher.verify(tag)
    return plaintext.decode("ascii")
  except:
    return False


msg=input("enter a message to be encrpt ")
nonce,ciphertext,tag=encrypt(msg)
plaintext = decrpyt(nonce,ciphertext,tag)
print(f'cipher text is {ciphertext}')

if not plaintext:
  print('message is corrupted')
else:
  print(f'plaintext is {plaintext}')

```
